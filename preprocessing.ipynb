```python
#remember to download the dataset automatically for the finished model

import csv
import numpy as np
import random

#source for data
URL = "https://www.cs.cornell.edu/people/pabo/movie-review-data/rotten_imdb.tar.gz"


with open('rotten_imdb/quote.tok.gt9.5000', encoding='latin-1') as f:
    subjective_text = [line.strip() for line in f.readlines()]


with open('rotten_imdb/plot.tok.gt9.5000', encoding='latin-1') as f:
    objective_text = [line.strip() for line in f.readlines()]

print("Subjective text length:", len(subjective_text))
print("Objective text length:", len(objective_text))

print("Subjective text sample:", subjective_text[0:5])
print("Objective text sample:", objective_text[0:5])

#shuffle the data
random.seed(42)
random.shuffle(subjective_text)
random.shuffle(objective_text)

#split the data into training, validation and testing sets 
train_subjective = subjective_text[:3200] 
val_subjective = subjective_text[3200:4000]
test_subjective = subjective_text[4000:]

train_objective = objective_text[:3200]
val_objective = objective_text[3200:4000]
test_objective = objective_text[4000:]

#merge the training and testing sets and create labels
train_data = train_subjective + train_objective
train_labels = [1] * len(train_subjective) + [0] * len(train_objective) # 1 for subjective, 0 for objective

val_data = val_subjective + val_objective
val_labels = [1] * len(val_subjective) + [0] * len(val_objective)

test_data = test_subjective + test_objective
test_labels = [1] * len(test_subjective) + [0] * len(test_objective)

#shuffle the training and testing sets
from sklearn.utils import shuffle
train_data, train_labels = shuffle(train_data, train_labels, random_state=42)
val_data, val_labels = shuffle(val_data, val_labels, random_state=42)
test_data, test_labels = shuffle(test_data, test_labels, random_state=42)

#save csv
with open('train_data.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['text', 'label'])
    for text, label in zip(train_data, train_labels):
        writer.writerow([text, label])

with open('test_data.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['text', 'label'])
    for text, label in zip(test_data, test_labels):
        writer.writerow([text, label])

with open('val_data.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['text', 'label'])
    for text, label in zip(val_data, val_labels):
        writer.writerow([text, label])
```
